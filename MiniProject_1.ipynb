```python
# MINI PROJECT — Customer Purchase Analyzer

import pandas as pd

df = pd.read_csv("purchases.csv")
df

valid_value = []
invalid_values = []

def clean_data(values):
    valid_values = []
    invalid_values = []

    for value in values:
        if pd.isna(value):
            invalid_values.append(value)
        else:
            try:
                num = float(value)
                if num >= 0:
                   valid_values.append(num)
                else: 
                    invalid_values.append(value)
            except:
                invalid_values.append(value)
            
    return valid_values, invalid_values

valid, invalid = clean_data(df["purchase_amount"])
print("Valid:", valid)
print("Invalid:", invalid)

def summarize(values):
    summary = {}
    
    summary["total_revenue"] = sum(values)
    summary["average_purchase"] = sum(values) / len(values)
    summary["max_purchase"] = max(values)
    summary["min_purchase"] = min(values)
    summary["count"] = len(values)

    return summary

summary = summarize(valid)
print(summary)

def generate_insights(summary):
    total = summary["total_revenue"]
    avg = summary["average_purchase"]
    max_val = summary["max_purchase"]
    min_val = summary["min_purchase"]
    count = summary["count"]

    insight = (
        f"Total revenue was {total} from {count} valid purchases."
        f"The average purchase value was {avg}."
        f"The highest purchase was {max_val} and the lowest {min_val}."
)

    return insight


print(generate_insights(summary))
```
